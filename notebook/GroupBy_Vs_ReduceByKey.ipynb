```python
# Imports 
from pyspark.sql import SparkSession
```

```python
#instanciate spark session
spark = SparkSession.builder\
        .appName('Aggregations')\
        .master("spark://172.22.0.2:7077")\
        .getOrCreate()
```

```python
spark.version
```

```python
sc = spark.sparkContext
```

```python
data = [('A',3), ('B',2), ('C',4), ('B',1), ('A',3), ('C',7), ('A',3)]
rdd = sc.parallelize(data*5000000)
rdd.take(10)
```

### 1. groupByKey : 

#### Exemple 1 

```python
import time 

start_time = time.time()

rdd_grouped = rdd.groupByKey().mapValues(lambda v : sum(v))
reduceByKey_duration = rdd_grouped.collect()

groupByKey_duration = time.time() - start_time
print(f'{groupByKey_duration} is the spend time')
```

#### Exemple 2

```python
data = [('A',3), ('B',2), ('C',4), ('B',1), ('A',3), ('C',7), ('A',3)]
rdd_data = sc.parallelize(data)
rdd_data.take(10)
```

```python
rdd_data_grouped = rdd_data.groupByKey().mapValues(list).collect()
print(rdd_data_grouped)
```

### 2. ReduceByKey : 

#### Exemple 1 

```python
start_time = time.time()

rdd_reduced = rdd.reduceByKey(lambda a,b : a+b )
reduceByKey_Result = rdd_reduced.collect()

groupByKey_duration = time.time() - start_time
print(f'{groupByKey_duration} is the spend time')
```
