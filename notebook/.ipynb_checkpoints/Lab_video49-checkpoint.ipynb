#### Phase 2: Data Loading 

##### What This Phase Does:

- Load CSV data with automatic schema inference

- Spark reads the header row and detects column types

- Verify data loaded successfully and preview schema

##### Why Schema Inference?

- Automatic column detection – matches CSV column order

- Type detection – Spark infers int, float, string types automatically

- Simpler code – no need to manually define 28 columns

##### Key Columns We'll Use:

Neighborhood, Battalion, SupervisorDistrict – Geographic hierarchy

##### SQL Equivalent:

-- Schema inference is like 

CREATE TABLE AS SELECT
CREATE TABLE fire_calls AS
SELECT * FROM csv_file('sf-fire-sample.csv', header=true, auto_detect=true);


````python
# Define the schema 


# Set up the data source file's path 
sf_fire_file = '/opt/spark-data/sf-fire-calls.csv'

# load dataframe with San Francesco fire data 

fire_df = spark.read\
    .option('header', 'true')\
    .option('inferSchema', 'true')\
    .csv(sf_fire_file, samplingRatio= 0.00001)


fire_df = fire_df.withColumn("Delay", col("Delay").cast("float"))

                                                        
print(' Data is loaded successfully:')
print(f'Columns : {len(fire_df.columns)}')
print('Schema : ')
fire_df.printSchema()

print('Sample data : ')
pdf = fire_df.limit(5).toPandas()
display(pdf)
````

#### Phase 3 : Data Enrishment

##### What This Phase Does:

- Parse CallDate string to proper date type

- Extract time dimensions: year, month, week

- Categorize call types into broader groups (Medical, Fire, Alarm, Other)

##### Why This Matters:

- Time dimensions enable weekly aggregations and year-over-year comparisons

- Call type categories simplify analysis (100+ specific types → 4 categories)

- Proper date types enable date arithmetic and window functions

SQL Equivalent:

```sql 
SELECT     
    *,    
    TO_DATE(CallDate, 'MM/dd/yyyy hh:mm:ss a') AS call_date,
    YEAR(call_date) AS call_year,
    MONTH(call_date) AS call_month,
    WEEK(call_date) AS call_week,
    CASE
        WHEN CallType LIKE '%Medical%' THEN 'Medical'
        WHEN CallType LIKE '%Fire%' THEN 'Fire'
        WHEN CallType LIKE '%Alarm%' THEN 'Alarm'
        ELSE 'Other'
    END AS call_type_category
FROM fire_calls;
```

##### Step 5.1: Parse Dates and Extract Time Dimensions 

````python
enriched_df = fire_df\
    .withColumn('call_date', to_date('CallDate', 'MM/dd/yyyy'))\
    .withColumn('call_year', year('call_date'))\
    .withColumn('call_month', month('call_date'))\
    .withColumn('call_week', weekofyear('call_date'))

print("Added time columns : call_date, call_year, call_month, and call_week")
pdf = enriched_df.limit(5).toPandas()
display(pdf)
````

##### Step 5.2: Categorize Call Types

````python
enriched_df = enriched_df\
    .withColumn(
        'call_type_category',
        when(col('CallType').contains('Medical'),'Medical')
        .when(col('CallType').contains('Fire'), 'Fire')
        .when(col('CallType').contains('Alarm'), 'Alarm')
        .otherwise('Other')
     )

print("Added call_type_categories column")

# preview categorization 

result= enriched_df.select('CallType', 'call_type_category')
pdf = result.limit(5).toPandas()
display(pdf)
````

#### Phase 4 : Weekly Aggregations 

##### Step 6.1: Basic Aggregations (Count, Average)

##### What This Phase Does:  
Group data by multiple dimensions and calculate comprehensive metrics:

- Grouping: Neighborhood + Battalion + District + CallType + Year + Week

- Metrics: Counts, averages, percentiles (p25, p50, p75, p95), standard deviation, conditional counts

##### Why This Matters:

- Weekly granularity balances detail vs. noise (daily = too noisy, monthly = loses trends)

- Percentiles show distribution (median more robust than average for skewed data)

- Standard deviation measures consistency (critical for “efficiency”)

- Multi-dimensional grouping enables comparisons at different levels

##### Business Value:

- Count: Volume capacity (can handle high call volume?)

- Average: Central tendency of response time

- Percentiles: Understand outliers (95th percentile = worst-case scenarios)

- Std Dev: Consistency (low = predictable performance)

SQL Equivalent:


SELECT

    Neighborhood, Battalion, SupervisorDistrict,
    call_type_category, call_year, call_week,
    COUNT(*) AS weekly_calls,
    AVG(Delay) AS avg_delay,
    PERCENTILE_APPROX(Delay, 0.25) AS delay_p25,
    PERCENTILE_APPROX(Delay, 0.50) AS delay_median,
    PERCENTILE_APPROX(Delay, 0.75) AS delay_p75,
    PERCENTILE_APPROX(Delay, 0.95) AS delay_p95,
    STDDEV(Delay) AS delay_stddev,
    SUM(CASE WHEN Priority = '3' THEN 1 ELSE 0 END) AS high_priority_calls
FROM enriched_calls
GROUP BY Neighborhood, Battalion, SupervisorDistrict,
         call_type_category, call_year, call_week;


````python
weekly_metrics = enriched_df.groupBy(
    'Neighborhood',
    'Battalion',
    'SupervisorDistrict',
    'call_type_category',
    'call_year',
    'call_week'
).agg(
    count('*').alias('weekly_calls'),
    avg('Delay').alias('avg_delay'),
    
)


print("Step 1 : Calculated basic aggregations (count, avg)")
print(f"Result : {weekly_metrics.count():,} rows")


result = weekly_metrics.orderBy('Weekly_calls')
pdf = result.limit(10).toPandas()
display(pdf)
````

##### Step 6.1: Add Percentile Calculattions 

![percentileVsPercentile_approx](./percentile.png )
 
##### Why We Choose PERCENTILE_APPROX: 




````python
# Exact percentile - NOT recommended for big data
# percentile(col, [0.25, 0.5, 0.75])  # Very expensive!

# Approximate percentile - RECOMMENDED ✓
expr("percentile_approx(Delay, 0.25)")  # Fast, scalable

# Can specify accuracy (default = 10000 = 99.9% accurate)
expr("percentile_approx(Delay, 0.25, 10000)")
````

##### Real-world example:

- Dataset: 1 billion emergency calls

- Exact percentile: 45 minutes, full cluster shuffle

- Approximate percentile: 3 minutes, 99.9% accurate

- Business decision: Same outcome, 15x faster!

````python
# Part 2 :Add percentile to understand delay distribution 

weekly_metrics = enriched_df.groupBy(
    'Neighborhood',
    'Battalion',
    'SupervisorDistrict',
    'call_type_category',
    'call_year',
    'call_week'
).agg(
    count('*').alias('weekly_calls'),
    avg('Delay').alias('avg_delay'),
    expr('percentile_approx(Delay, 0.25)').alias('Delay_p25'),
    expr('percentile_approx(Delay, 0.5)').alias('Delay_median'),
    expr('percentile_approx(Delay, 0.75)').alias('Delay_p75'),
    expr('percentile_approx(Delay, 0.95)').alias('Delay_p95')
    
)

print("Step 2 : Added percentiles (p25, p50, p75, p95)")

# Preview percentile distribution 

result = weekly_metrics.select(
    'call_year',
    'Neighborhood',
    'call_type_category',
    'call_week',
    'weekly_calls',
    'avg_delay',
    'Delay_p25',
    'Delay_median',
    'Delay_p75',
    'Delay_p95',
).orderBy('call_year','call_week','Neighborhood','call_type_category', )

pdf = result.limit(10).toPandas()
display(pdf)
````

````python
# Part 2 :Complete aggregation with all metrics

weekly_metrics = enriched_df.groupBy(
    'Neighborhood',
    'Battalion',
    'SupervisorDistrict',
    'call_type_category',
    'call_year',
    'call_week'
).agg(
    # basic metrics 
    count('*').alias('weekly_calls'),
    avg('Delay').alias('avg_delay'),

    # Distribution (percentiles)
    
    expr('percentile_approx(Delay, 0.25)').alias('Delay_p25'),
    expr('percentile_approx(Delay, 0.5)').alias('Delay_median'),
    expr('percentile_approx(Delay, 0.75)').alias('Delay_p75'),
    expr('percentile_approx(Delay, 0.95)').alias('Delay_p95'),
    
    # Consistency metrics
    stddev('Delay').alias('Delay_stddev'),

    # Conditional count (high priority Calls)

    sum(when(col("Priority") == '3', 1).otherwise(0)).alias('high_priority_calls')
    
)

result = weekly_metrics.select(
    # dimensions (time, georaphic, and call_category)
    'call_year', 
    'call_week',  
    'Neighborhood',   
    'call_type_category', 
    
    # metrics  
    'weekly_calls',
    'high_priority_calls',
    'avg_delay',
    'Delay_stddev',
    'Delay_p25',
    'Delay_median',
    'Delay_p75',
    'Delay_p95',
).orderBy('call_year','Neighborhood','call_type_category', 'call_week')

pdf = result.limit(10).toPandas()
display(pdf)


print("Step 3 : Added stddev and conditional count")
print(f"Final Metrics : {weekly_metrics.count():,} rows")

````

##### Step 6.4: Preview Weekly Metrics 

Semilar to SQL :

SELECT

```

````python
result = weekly_metrics.orderBy(desc('weekly_calls'))
pdf = result.limit(10).toPandas()

print('Top 10 neighborhoods by weekly call volume :')
display(pdf)
````

#### Phase 5: Comparative Analysis with Window Functions

##### What This Phase Does:

- Battalion comparison: How does each neighborhood compare to others in the same battalion?

- Moving averages: What’s the 7‑week trend (smooth out weekly noise)?

- Year‑over‑year: Is performance improving compared to last year?

##### Why Window Functions?:

- Window functions calculate values across related rows without collapsing groups.

- PARTITION BY = divide data into groups

- ORDER BY = order within groups

- OVER() = apply function across the window

- Unlike GROUP BY, window functions keep all original rows while adding comparative metrics.

##### Business Value:

- Battalion comparison: Fair comparison (same geographic/operational conditions)

- Moving averages: Identify trends vs. random weekly fluctuations

- YoY comparison: Track improvement initiatives, seasonal patterns

SQL Equivalent:

-- Battalion comparison

```sql

SELECT *,
    AVG(avg_delay) OVER (
        PARTITION BY Battalion, call_type_category, call_year
    ) AS battalion_avg_delay,

-- Moving average (7 weeks)
    AVG(avg_delay) OVER (
        PARTITION BY Neighborhood, call_type_category
        ORDER BY call_week
        ROWS BETWEEN 6 PRECEDING AND CURRENT ROW
    ) AS moving_avg_7w,
-- Year-over-year
    LAG(avg_delay, 1) OVER (
        PARTITION BY Neighborhood, call_type_category, call_week
        ORDER BY call_year
    ) AS prev_year_delay
FROM weekly_metrics;


##### Step 7.1: Define Battalion window  

````python
from pyspark.sql import Window

# Define Window : group neighborhoods by their battallion
# This creates 'partitions' for fair comparison
battallion_window = Window.partitionBy(
    'Battalion',
    'call_type_category',
    'call_year',
)

print("Defined battallion window for comparison")
print("Partition by : Battalion, call_year ,and call_type_category")
print("Purpose : Compare neighborhoods within same battalion/conditions")
````

##### Step 7.2: Calculate Battalion Average

Apply window function to calculate average delay across battalion.

````python
with_battalion_avg = weekly_metrics.withColumn(
    'battalion_avg_delay',
    avg('avg_delay').over(battallion_window)
)

print("Added battalion_avg_delay column")
print("each row now shows its battalion's average delay in a year")
````

##### Step 7.3: Calculate Differnce form Battalion Average

- Positive value = slower than battalion average.
- Negative = faster 

##### Step 7.4: Calculate 7-week Moving Average

Moving averages smooth out weekly noise to see true trends.

###### Similar to SQL: 
```sql 
AVG(AVG_DELAY)
    OVER(
        PARTITION BY NEIGHBORHOOD 
        ORDER BY WEEK 
        ROWS BETWEEN 6 PRECEIDING AND CURRENT ROW
    )
```

````python
moving_avg_window = Window\
    .partitionBy('Neighborhood', 'call_type_category')\
    .orderBy('call_year','call_week')\
    .rowsBetween(-6, 0)

# Calculate 7 week rolling average 
with_moving_avg = with_battalion_avg.withColumn(
    'moving_avg_7w',
    avg('avg_delay').over(moving_avg_window)
)

print('Added 7-week moving average')

#Preview moving averages

result = with_moving_avg.select(
    'call_year','Battalion','Neighborhood','call_type_category', 'call_week', 'avg_delay', 'moving_avg_7w'
).limit(20).toPandas()

display(result)
````

##### Step 7.5: Year-over-Year Analysis - Define Window and Get Previous Year 


Compare same week this year vs last year to track imporvement.
Similat to SQL: LAG(avg_delay, 1) OVER(PARTITION BY Neighborhood, call_week ORDER BY call_year)

````python
YoY_window = Window\
    .partitionBy('Neighborhood','call_type_category','call_week')\
    .orderBy('call_year')  

with_yoy= with_moving_avg.withColumn(
    'prev_year_delay', lag('avg_delay',1).over(YoY_window)
)

result = with_yoy.select(
    'Neighborhood','call_type_category', 'call_year', 'call_week', 'avg_delay', 'prev_year_delay'  
).orderBy('Neighborhood','call_type_category', 'call_year', 'call_week')

pdf = result.limit(10).toPandas()
display(pdf)
````

##### Step 7.6: Calculate the YoY Growth

Negataive growth : good imporvement (delays are decreasing)

````python
with_yoy = with_yoy.withColumn(
    'yoy_growth',
    when(
        col('prev_year_delay').isNotNull(), 
        (((col('avg_delay')- col('prev_year_delay'))/ col('prev_year_delay'))*100)
    ).
    otherwise(None)
)


print('Added the yoy growth column')
print('Negative % = imporvement ')
print('Positive % = degradation ')


# Preview yoy comparison 

result = with_yoy.filter(col('prev_year_delay').isNotNull()).select(
    'Neighborhood','call_type_category', 'call_year', 'call_week', 'avg_delay', 'prev_year_delay', 'yoy_growth'
).orderBy('Neighborhood','call_type_category', 'call_year','call_week')

pdf = result.limit(10).toPandas()
display(pdf)

````

#### Phase 6: Efficiency Scoring & Ranking

##### 🔹 What This Phase Does
1. Create custom efficiency score combining multiple factors  
2. Rank neighborhoods within each district using 4 different ranking methods  
3. Filter to top 3 performers per district and call type  

---

##### 🔹 The Efficiency Formula
**Efficiency Score =**  
- (100 - avg_delay) × 0.40 → Fast response (40% weight)  
-  +(weekly_calls) × 0.30 → High volume capacity (30% weight)  
-  -(delay_stddev) × 0.30 → Low variation / consistency (30% weight)  

---

##### 🔹 Why This Matters
- **Multi-factor scoring:** Single metric (e.g., just speed) can be misleading  
  - Fast but inconsistent? Not truly efficient  
  - Fast but low volume? Can’t handle demand  
- **Weighted components:** Business priority (speed matters most)  
- **4 ranking methods:** Different tie-handling for different use cases  

---

##### 🔹 Ranking Methods Explained

| Method         | Ties Behavior     | Example       | Use Case                  |
|----------------|-------------------|---------------|---------------------------|
| **rank()**     | Gaps after ties   | 1, 2, 2, 4    | Competition ranking        |
| **dense_rank()** | No gaps         | 1, 2, 2, 3    | Level/tier assignment      |
| **row_number()** | Always unique   | 1, 2, 3, 4    | Strict top-N filtering     |
| **percent_rank()** | 0.0 to 1.0    | 0.0, 0.33, 0.33, 1.0 | Percentile position |

---

##### 🔹 SQL Equivalent
```sql
-- Example SQL for efficiency scoring & ranking
SELECT

    *,
    --custom score
    (100 - avg_delay) * 0.40
    + (weekly_calls) * 0.30
    - (delay_stddev) * 0.30 AS efficiency_score,

    -- Ranking
    RANK() OVER (PARTITION BY District ORDER BY efficiency_score DESC) AS rank,
    DENSE_RANK() OVER (PARTITION BY District ORDER BY efficiency_score DESC) AS rank,    
    ROW_NUMBER() OVER (PARTITION BY District ORDER BY efficiency_score DESC) AS rank,
    PERCENT_RANK() OVER (PARTITION BY District ORDER BY efficiency_score DESC) AS rank,
FROM with_yoy
    
WHERE rank <= 3 ;  -- top 3 only 

```



##### Step 6.1: Calculate Efficiency Score 

````python
efficiency_df = with_yoy.withColumn(
    'efficiency_score',

    #component 1: Fast Response 
    (lit(100)- col("avg_delay")) * 0.4 +
    
    #component 2: Fast Response 
    (col("weekly_calls")) * 0.3 -   # Can handle many calls 
    
    #component 3: Low variation  
    (col("delay_stddev")) * 0.3    # Conssistent performance
    
)



print('Added the effeciency column')


# Preview yoy comparison 

result = efficiency_df\
    .filter(col('delay_stddev').isNotNull())\
    .select(
        'call_year','Neighborhood','call_type_category','call_week', # dimesions
         'avg_delay', 'weekly_calls', 'delay_stddev',  # criterias 
        'efficiency_score' # the decision score 
    ).orderBy(
        'call_year',
        'efficiency_score'
    )\
    .filter('call_year = 2018')\


pdf = result.limit(10).toPandas()
display(pdf)
````

````python
# Define ranking window and aply all 4 ranking methods 
# Important : /ranking are calculated for all districts (1-11)
# The partionBy groups data, but processe all partitions 

ranking_window = Window \
    .partitionBy("SupervisorDistrict", "call_type_category", "call_year") \
    .orderBy(desc("efficiency_score"))

# Apply all four ranking methods to the entire dataset 
# Each district gets its own rankings (partition), buta all districrs are ranked 

ranked_df = efficiency_df \
    .withColumn("rank", rank().over(ranking_window)) \
    .withColumn("dense_rank", dense_rank().over(ranking_window)) \
    .withColumn("row_number", row_number().over(ranking_window)) \
    .withColumn("percent_rank", percent_rank().over(ranking_window))




print('Added 4 ranking columns')
print(f'Ranking applied to all {efficiency_df.select("SupervisorDistrict").distinct().count()} districts.')

# Preview ranking from q only for readability 
# NOTE : This filtre is only for display - actual rankings include all districts!

result = ranked_df.select(
    "SupervisorDistrict", "Neighborhood", "call_type_category",
    "efficiency_score", "rank", "dense_rank", "row_number"
).filter(col("SupervisorDistrict") == "1") \
 .orderBy("call_type_category", "row_number")


pdf = result.limit(10).toPandas()
print("\n Sample from District 1 (ranking calculated for all districrs); ")
display(pdf)

````

##### Step 6.2: Rank Neighborhoods (4 Different Ranking Methods)

**Educational Note** : Why All 4 Methods?
For our business requirement (top 3), we only need `row_number()`.  
But showing all 4 demonstrates:
- Different tie-handling behaviors (common interview question!)
- When to use each method for different business scenarios

---

🔹 **Comparison of Ranking Methods**

- **rank()** → Competition ranking (1, 1, 3) – gaps after ties  
- **dense_rank()** → Level assignment (1, 1, 2) – no gaps  
- **row_number()** → Top-N filtering (1, 2, 3) – always unique ← *We use this!*  
- **percent_rank()** → Percentile position (0.0 to 1.0)  


````python
top_performers = ranked_df.filter((col("row_number") <= 3) & (col("supervisorDistrict").isNotNull()) & (col("Neighborhood").isNotNull()))

print(f"Filtered to top 3: {top_performers.count()} rows")
print("Top performers per district and call type:")

result = top_performers.select(
    "call_year","SupervisorDistrict", "call_type_category", "Neighborhood",
    "row_number", "efficiency_score", "avg_delay", "weekly_calls"
).orderBy("call_year", "SupervisorDistrict", "call_type_category", "row_number")

display(result.limit(20).toPandas())
````

#### Phase 7: Hierarchical Aggregation & Performance Tiers 
##### Step: Hierarchical Rollups & Performance Tiers

🔹 **What This Phase Does**
1. Apply `rollup()` to create hierarchical subtotals at multiple levels  
2. Identify which rollup level each row represents  
3. Assign performance tiers based on efficiency and delay thresholds  

---

🔹 **Rollup Hierarchy**
1. **Neighborhood detail** (most granular)  
   - rollup  
2. **District + CallType totals**  
   - rollup  
3. **District totals**  
   - rollup  
4. **Grand total** (all data)  

---

🔹 **Why Rollup?**
- **Executive reporting:** Different stakeholders need different levels  
  - Fire Chief → Grand totals and district summaries  
  - Battalion Chief → Category totals within their district  
  - Station Captain → Neighborhood details  
- **Automatic subtotals:** One query generates all levels (vs. multiple queries)  
- **NULL indicators:** NULL values show which dimension was rolled up  

---

🔹 **Performance Tiers**

| Tier | Criteria | Meaning |
|------|-----------|---------|
| **Tier 1: Excellent** | efficiency > 80 AND delay < 5 | Exceptional performance |
| **Tier 2: Good** | efficiency > 60 AND delay < 8 | Above average |
| **Tier 3: Average** | efficiency > 40 | Meets baseline |
| **Tier 4: Needs Improvement** | Otherwise | Below standards |

---

🔹 **SQL Equivalent**
```sql
-- Example SQL for rollup and tier assignment
-- Rollup creates automatic subtotals
SELECT
    SupervisorDistrict,
    call_type_category,
    Neighborhood,
    SUM(weekly_calls) AS total_calls,
    AVG(avg_delay) AS average_delay,
    AVG(efficiency_score) AS avg_efficiency,
    CASE
        WHEN SupervisorDistrict IS NULL THEN 'GRAND_TOTAL'
        WHEN call_type_category IS NULL THEN 'DISTRICT_TOTAL'
        WHEN Neighborhood IS NULL THEN 'CATEGORY_TOTAL'
        ELSE 'NEIGHBORHOOD_DETAIL'
    END AS aggregation_level,
    CASE
        WHEN avg_efficiency > 80 AND average_delay < 5 THEN 'Tier 1: Excellent'
        WHEN avg_efficiency > 60 AND average_delay < 8 THEN 'Tier 2: Good'
        WHEN avg_efficiency > 40 THEN 'Tier 3: Average'
        ELSE 'Tier 4: Needs Improvement'
    END AS performance_tier
FROM top_performers
GROUP BY ROLLUP(SupervisorDistrict, call_type_category, Neighborhood);

```


##### Step 7.1: Create Hierarchical Summary with ROLLUP 

````python
# rollup() - Creates hierarchical subtotals
# Generates totals in multiple levels of aggregation
# NULL values indicate which level of rollup

print("Creating rollup (this calculates many totals...)")

rollup_result = top_performers.rollup(
    "SupervisorDistrict",  # Level 1
    "call_type_category",  # Level 2
    "Neighborhood"         # Level 3 (detail)
).agg(
    sum("weekly_calls").alias("total_calls"),       # sum() - Add up all values
    avg("avg_delay").alias("average_delay"),        # avg() - Average delay
    avg("efficiency_score").alias("avg_efficiency"),# avg() - Average efficiency
    count(when(col("yoy_growth") < 0, 1)).alias("improving_weeks") # Conditional count
)

# when() - Conditional logic to identify rollup level
rollup_with_level = rollup_result.withColumn("aggregation_level",
    when(col("SupervisorDistrict").isNull(), "GRAND_TOTAL")   # All data
    .when(col("call_type_category").isNull(), "DISTRICT_TOTAL") # District level
    .when(col("Neighborhood").isNull(), "CATEGORY_TOTAL")       # Category level
    .otherwise("NEIGHBORHOOD_DETAIL")                           # Detail level
)

print("✓ Created hierarchical rollup")

result = rollup_with_level.groupby("aggregation_level").count()
display(result.limit(20).toPandas())

# Open 'pdf' in Data Wrangler
````

##### Step 7.2: Assign Performance Tiers 

Categorize performance based on efficiency score and average delay thresholds

---

````python
tiered_result = rollup_with_level.withColumn("performance_tier",
    when((col("avg_efficiency") > 80) & (col("average_delay") < 5),
         "Tier 1: Excellent")
    .when((col("avg_efficiency") > 60) & (col("average_delay") < 8),
          "Tier 2: Good")
    .when(col("avg_efficiency") > 40,
          "Tier 3: Average")
    .otherwise("Tier 4: Needs Improvement")
)

print("✓ Added performance tiers")
print("\nPerformance tier distribution:")

result = tiered_result.groupBy("performance_tier").count().orderBy(desc("count"))
pdf = result.limit(20).toPandas()
display(pdf)
````
