```python
from pyspark.sql import SparkSession 
```

```python
#create a session 
spark = SparkSession.builder \
    .master("spark://spark-master:7077") \ 
    .appName("firstSparkApplication") \
    .getOrCreate()
```

```python
df = spark.read.csv("/opt/spark-data/tips.csv", header=True, inferSchema=True)
```

```python
df.show()
```

```python
df.columns 
```

```python
df2 = df.select(*['total_bill', 'tip', 'sex', 'smoker', 'day', 'time', 'size']).limit(5)
```

```python
df2.show()
```

```python
df_male = df.where("sex = 'Male'")
df_male.show()
```
