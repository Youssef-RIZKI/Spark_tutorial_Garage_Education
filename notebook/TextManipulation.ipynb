```python
from pyspark.sql import SparkSession
```

```python
# instantiate the sparkSession
spark = SparkSession.builder\
        .appName("TextManipulation")\
        .master("spark://172.22.0.2:7077")\
        .getOrCreate()
```

```python
spark.version
```

```python
sc = spark.sparkContext
```

```python
text = ['hello Mouad!!', 'Hello Ismail!!', 'Hello Youssef!!' ]
text_rdd = sc.parallelize(text)
text_rdd.take(10)
```

```python
rdd_flat_map = text_rdd.flatMap(lambda x : x.split(' '))
rdd_flat_map.take(10)
```

```python
rdd_map = text_rdd.map(lambda x : x.split(' '))
rdd_map.take(10)
```

```python
rdd_upper = rdd_flat_map.map (lambda w : w.upper())
rdd_upper.take(10)
```

```python
rdd_lower = rdd_flat_map.map (lambda w : w.lower())
rdd_lower.take(10)
```

```python
spark.stop()
```
